
# Canny Edge Detection using OpenCV

## Aim
To implement the Canny Edge Detection algorithm on a sample image and analyze the effect of different parameter settings.


```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Load image
image = cv2.imread('sample_image.png')

# Convert to RGB for display
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,4))
plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Apply Gaussian Blur
blurred = cv2.GaussianBlur(gray, (5,5), 1)

plt.figure(figsize=(6,4))
plt.imshow(blurred, cmap='gray')
plt.title('Blurred Grayscale Image')
plt.axis('off')
plt.show()
```

```python

# Apply Canny Edge Detection
edges = cv2.Canny(blurred, 50, 150)

plt.figure(figsize=(6,4))
plt.imshow(edges, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')
plt.show()
```


## Parameter Analysis

Different threshold values produce different edge outputs.
- Lower thresholds detect more edges and noise.
- Higher thresholds detect only stronger edges.


```python

# Different parameter settings

edges1 = cv2.Canny(blurred, 30, 100)
edges2 = cv2.Canny(blurred, 100, 200)

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(edges1, cmap='gray')
plt.title('Thresholds: 30, 100')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(edges2, cmap='gray')
plt.title('Thresholds: 100, 200')
plt.axis('off')

plt.show()
```


# Conclusion

The Canny Edge Detection algorithm effectively detects object boundaries in images.

### Observations
- Lower threshold values produce more detailed edges but may include noise.
- Higher threshold values remove weak edges and keep only strong boundaries.
- Gaussian blur helps reduce noise before edge detection.
